### 2. Missing Values Analysis

Next, let's identify and handle any missing values in the dataset.

```python
# Check for missing values
missing_values = df.isnull().sum()
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_info = pd.DataFrame({
    'Missing Count': missing_values,
    'Missing Percentage': missing_percentage
})

display(missing_info[missing_info['Missing Count'] > 0].sort_values(by='Missing Percentage', ascending=False))

# Example of handling missing values (if any)
# For numerical columns, you might impute with mean, median, or mode.
# For categorical columns, you might impute with mode or a 'Unknown' category.
# For 'fare_amount' or 'passenger_count', it's often better to drop rows if missing, as they are critical.

# Drop rows where 'fare_amount' is missing (if applicable, as it's a target variable)
if 'fare_amount' in df.columns and df['fare_amount'].isnull().any():
    initial_rows = df.shape[0]
    df.dropna(subset=['fare_amount'], inplace=True)
    print(f"Dropped {initial_rows - df.shape[0]} rows with missing 'fare_amount'.")

# For 'passenger_count', if missing, it might be reasonable to impute with the mode or drop, depending on context.
# Let's impute with mode as an example if there are missing values and it's not a dummy df.
if 'passenger_count' in df.columns and df['passenger_count'].isnull().any():
    mode_passenger_count = df['passenger_count'].mode()[0]
    df['passenger_count'].fillna(mode_passenger_count, inplace=True)
    print(f"Imputed missing 'passenger_count' with mode: {mode_passenger_count}")

# After handling, recheck for missing values
print("\nMissing values after initial handling:")
display(df.isnull().sum())
```
