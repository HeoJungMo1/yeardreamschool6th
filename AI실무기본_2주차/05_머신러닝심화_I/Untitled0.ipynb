```python
from pycaret.datasets import get_data
from pycaret.classification import ClassificationExperiment

# 데이터 로드
data = get_data('iris')

# 실험 초기화
exp = ClassificationExperiment()
exp.setup(data, target='species', session_id=123, verbose=False) # verbose=False로 깔끔하게 확인

# 모델 비교
best_model = exp.compare_models()
print("성공적으로 최적 모델을 찾았습니다:", type(best_model))
```
