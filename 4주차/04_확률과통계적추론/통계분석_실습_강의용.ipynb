# 문제 1 - 독립표본 t-검정

## [문제 1] 온라인 vs 오프라인 채널 영업 사원 월매출 비교 *(세일즈·마케팅)*

대형 유통기업 세일즈팀에서 온라인 채널과 오프라인 채널 영업 사원을 각각 80명, 75명 무작위 선발하여 **지난달 월매출(만 원)**을 수집하였다.  
두 채널 간 월매출에 유의미한 차이가 있는지 **독립표본 t-검정**으로 분석하고자 한다.

**[데이터]**
- 파일명: `sales_1_ttest_indep.csv`

**[컬럼 설명]**
- `emp_id` : 직원 식별 번호
- `channel` : 채널 구분 (온라인 / 오프라인)
- `monthly_sales` : 월매출 (만 원, 실수형)

**[문항]**
(1) 온라인·오프라인 각각의 월매출 **평균**을 구하시오. (소수점 셋째 자리까지 반올림)
(2) Levene 검정으로 등분산 여부를 판단한 뒤, 적절한 독립표본 t-검정을 실시하여 **검정통계량(t)**과 **p-value**를 구하시오. (소수점 셋째 자리까지 반올림)
(3) 유의수준 5%에서 귀무가설의 **기각/채택** 여부를 결정하시오.

---
## 독립표본 t-검정 핵심 정리

- **목적** : 두 독립 집단의 평균 차이가 통계적으로 유의한지 검정

- **가설**
  - H0 : 두 채널의 월매출 평균은 같다 (mu_online = mu_offline)
  - H1 : 두 채널의 월매출 평균은 다르다

- **등분산 검정 (Levene)**
  - p >= 0.05 : 등분산 가정 → `equal_var=True`
  - p < 0.05  : 이분산 → Welch t-test (`equal_var=False`)

- **p-value 해석 (alpha=0.05)**
  - p < 0.05  → H0 기각 → 두 채널 평균 차이 있음
  - p >= 0.05 → H0 채택 → 유의한 차이 없음

```python
import numpy as np
import pandas as pd
from scipy import stats

df = pd.read_csv(DATA_PATH + "sales_1_ttest_indep.csv", encoding="utf-8-sig")
print(df.head())

# 풀이
online  = df[df["channel"] == "온라인"]["monthly_sales"]
offline = df[df["channel"] == "오프라인"]["monthly_sales"]

# (1) 각 집단 평균
mean_on  = round(float(online.mean()), 3)
mean_off = round(float(offline.mean()), 3)
print("(1) 온라인 평균  :", mean_on, "만 원")
print("(1) 오프라인 평균:", mean_off, "만 원")

# (2) Levene 등분산 검정 후 t-검정
lev_stat, lev_p = stats.levene(online, offline)
print("\nLevene 검정 p-value:", round(float(lev_p), 3))
equal_var = lev_p >= 0.05
print("등분산 가정:", equal_var)

t_stat, p_val = stats.ttest_ind(online, offline, equal_var=equal_var)
ans2_t = round(float(t_stat), 3)
ans2_p = round(float(p_val), 3)
print("\n(2) t-통계량:", ans2_t)
print("(2) p-value  :", ans2_p)

# (3) 귀무가설 판정
decision = "기각" if ans2_p < 0.05 else "채택"
print("\n(3) 귀무가설 (H0: 두 채널 월매출 평균 동일) ->", decision)
```

# 문제 2 - 대응표본 t-검정

## [문제 2] 신규 디지털 마케팅 캠페인 전·후 전환율 비교 *(세일즈·마케팅)*

이커머스 기업 마케팅팀에서 60개 광고 소재에 대해 **리뉴얼 캠페인 전·후 전환율(%)**을 측정하였다.  
캠페인이 전환율 향상에 효과가 있는지 **대응표본 t-검정**으로 분석하고자 한다.

**[데이터]**
- 파일명: `sales_2_ttest_paired.csv`

**[컬럼 설명]**
- `ad_id` : 광고 소재 식별 번호
- `category` : 광고 카테고리 (패션 / 식품 / 전자기기)
- `pre_cvr` : 캠페인 전 전환율 (%)
- `post_cvr` : 캠페인 후 전환율 (%)

**[문항]**
(1) 캠페인 전·후 전환율의 **평균**과 **차이(후 - 전)의 평균**을 구하시오. (소수점 셋째 자리까지 반올림)
(2) 대응표본 t-검정을 실시하여 **검정통계량(t)**과 **p-value**를 구하시오. (소수점 셋째 자리까지 반올림)
(3) 유의수준 5%에서 캠페인이 전환율 향상에 효과가 있다고 할 수 있는지 **기각/채택** 여부를 결정하시오.

---
## 대응표본 t-검정 핵심 정리

- **목적** : 동일 대상의 두 측정값(전·후) 간 평균 차이 검정

- **가설** (단측: 캠페인 후 전환율 향상 기대)
  - H0 : 차이 없음 (mu_d = 0)
  - H1 : 캠페인 후 전환율이 더 높음 (mu_d > 0)

- **`scipy.stats.ttest_rel(post, pre, alternative='greater')`** 사용

```python
import numpy as np
import pandas as pd
from scipy import stats

# 데이터 생성
df = pd.read_csv(DATA_PATH + "sales_2_ttest_paired.csv", encoding="utf-8-sig")

# 풀이
pre  = df["pre_cvr"]
post = df["post_cvr"]
diff = post - pre

# (1) 평균 및 차이 평균
print("(1) 캠페인 전 전환율 평균   :", round(float(pre.mean()), 3), "%")
print("(1) 캠페인 후 전환율 평균   :", round(float(post.mean()), 3), "%")
print("(1) 차이(후-전) 평균        :", round(float(diff.mean()), 3), "%p")

# (2) 대응표본 t-검정 (단측: 후 > 전)
t_stat, p_val = stats.ttest_rel(post, pre, alternative="greater")
ans2_t = round(float(t_stat), 3)
ans2_p = round(float(p_val), 3)
print("\n(2) t-통계량:", ans2_t)
print("(2) p-value  :", ans2_p)

# (3) 귀무가설 판정
decision = "기각" if ans2_p < 0.05 else "채택"
print("\n(3) 귀무가설 (H0: 캠페인 전환율 향상 효과 없음) ->", decision)
if decision == "기각":
    print("    -> 리뉴얼 디지털 캠페인은 전환율 향상에 유의미한 효과가 있다.")
```

# 문제 3 - 일원배치 분산분석 (One-Way ANOVA)

## [문제 3] 직군별 신입 공채 지원자 토익 점수 비교 *(HR)*

대기업 인사팀에서 영업직·마케팅직·IT직·기획직 4개 직군 신입 공채 지원자를 각 40명씩 무작위 추출하여 **토익 점수**를 수집하였다.  
직군에 따라 토익 점수에 차이가 있는지 **일원배치 분산분석(One-Way ANOVA)**으로 검정하고자 한다.

**[데이터]**
- 파일명: `hr_3_anova.csv`

**[컬럼 설명]**
- `applicant_id` : 지원자 식별 번호
- `job_type` : 직군 (영업직 / 마케팅직 / IT직 / 기획직)
- `toeic_score` : 토익 점수 (정수, 단위: 점)

**[문항]**
(1) 각 직군별 토익 점수의 **평균**을 구하시오. (소수점 셋째 자리까지 반올림)
(2) 일원배치 ANOVA를 실시하여 **F-통계량**과 **p-value**를 구하시오. (소수점 셋째 자리까지 반올림)
(3) 유의수준 5%에서 귀무가설의 **기각/채택** 여부를 결정하시오.
(4) (3)에서 귀무가설을 기각한 경우, Tukey HSD 사후검정을 실시하여 **유의미하게 차이나는 직군 쌍**을 모두 구하시오.

---
## One-Way ANOVA 핵심 정리

- **목적** : 3개 이상 집단의 평균 차이를 동시에 검정

- **가설**
  - H0 : 모든 직군의 토익 평균이 같다
  - H1 : 적어도 하나의 직군 평균이 다르다

- **사후검정** : ANOVA 기각 시에만 실시 → Tukey HSD

```python
import numpy as np
import pandas as pd
from scipy import stats
from statsmodels.stats.multicomp import pairwise_tukeyhsd


df = pd.read_csv(DATA_PATH + "hr_3_anova.csv", encoding="utf-8-sig")
print(df.head())

# 풀이
groups = df.groupby("job_type")["toeic_score"]

# (1) 직군별 평균
print("(1) 직군별 토익 점수 평균:")
for name, grp in groups:
    print(f"    {name}: {round(float(grp.mean()), 3)} 점")

# (2) One-Way ANOVA
group_data = [g.values for _, g in groups]
f_stat, p_val = stats.f_oneway(*group_data)
ans2_f = round(float(f_stat), 3)
ans2_p = round(float(p_val), 3)
print("\n(2) F-통계량:", ans2_f)
print("(2) p-value  :", ans2_p)

# (3) 귀무가설 판정
decision = "기각" if ans2_p < 0.05 else "채택"
print("\n(3) 귀무가설 (H0: 모든 직군 토익 평균 동일) ->", decision)

# (4) Tukey HSD 사후검정
if decision == "기각":
    tukey = pairwise_tukeyhsd(
        endog=df["toeic_score"],
        groups=df["job_type"],
        alpha=0.05
    )
    print("\n(4) Tukey HSD 사후검정 결과:")
    print(tukey.summary())
    result_df = pd.DataFrame(
        data=tukey._results_table.data[1:],
        columns=tukey._results_table.data[0]
    )
    sig_pairs = result_df[result_df["reject"] == True][["group1", "group2"]]
    print("유의미하게 차이나는 직군 쌍:")
    for _, row in sig_pairs.iterrows():
        print(f"    {row['group1']} vs {row['group2']}")
else:
    print("    -> 사후검정 불필요 (귀무가설 채택)")
```

# 문제 4 - 카이제곱 검정 (Chi-Square Test)

## [문제 4] 채용 전형 유형과 최종 합격 여부의 독립성 검정 *(HR)*

채용 컨설팅 기업에서 공개채용·추천채용·인턴전환 세 가지 전형을 거친 지원자 300명의 최종 합격 여부를 조사하였다.  
채용 전형 유형과 최종 합격 여부 사이에 연관성이 있는지 **카이제곱 독립성 검정**으로 분석하고자 한다.

**[데이터]**
- 파일명: `hr_4_chisq.csv`

**[컬럼 설명]**
- `applicant_id` : 지원자 식별 번호
- `hire_type` : 채용 전형 유형 (공개채용 / 추천채용 / 인턴전환)
- `final_pass` : 최종 합격 여부 (합격 / 불합격)

**[문항]**
(1) 채용 전형별 합격/불합격 **교차표(contingency table)**를 작성하시오.
(2) 카이제곱 검정을 실시하여 **검정통계량(chi2)**과 **p-value**를 구하시오. (소수점 셋째 자리까지 반올림)
(3) 유의수준 5%에서 귀무가설의 **기각/채택** 여부를 결정하고, 연관성 해석을 서술하시오.

---
## 카이제곱 독립성 검정 핵심 정리

- **목적** : 두 범주형 변수 간 독립성(연관성) 검정

- **가설**
  - H0 : 채용 전형 유형과 합격 여부는 독립이다 (연관성 없음)
  - H1 : 두 변수는 독립이 아니다 (연관성 있음)

- **조건** : 기대도수 >= 5 (대부분의 셀에서)
  + 5미만 => 피셔의 정확도(Fisher Exact Test) 검정
- **`scipy.stats.chi2_contingency(ct)`** 사용

```python
import numpy as np
import pandas as pd
from scipy.stats import chi2_contingency

# 데이터 생성
df = pd.read_csv(DATA_PATH + "hr_4_chisq.csv", encoding="utf-8-sig")
print(df.head())

# 풀이
# (1) 교차표
ct = pd.crosstab(df["hire_type"], df["final_pass"])
print("(1) 교차표 (채용 전형 x 합격 여부):")
print(ct)

# (2) 카이제곱 검정
chi2, p, dof, expected = chi2_contingency(ct)
ans2_chi2 = round(float(chi2), 3)
ans2_p    = round(float(p), 3)
print(f"\n(2) chi2-통계량 : {ans2_chi2}")
print(f"(2) 자유도      : {dof}")
print(f"(2) p-value     : {ans2_p}")
print("\n기대도수 행렬:")
print(pd.DataFrame(expected, index=ct.index, columns=ct.columns).round(2))

# (3) 귀무가설 판정
decision = "기각" if ans2_p < 0.05 else "채택"
print(f"\n(3) 귀무가설 (H0: 채용 전형과 합격 여부 독립) -> {decision}")
if decision == "기각":
    print("    -> 채용 전형 유형과 최종 합격 여부는 통계적으로 유의미한 연관성이 있다.")
else:
    print("    -> 채용 전형 유형과 최종 합격 여부는 통계적으로 독립적이다.")
```

**[문항]**
(1) `call_count`, `visit_count`, `work_years`, `training_score`를 독립변수로 하는 다중회귀모형을 적합하고, **회귀계수가 가장 높은 변수**를 구하시오.

(3) 회귀모형의 **결정계수(R²)**와 **수정된 결정계수(Adj. R²)**를 구하시오. (소수점 셋째 자리까지 반올림)
---

## 가설검정
- H0 = 기울기가 존재하지 않는다(=기울기가 0이다)
- H1 = 기울기가 존재한다 (=기울기가 0이 아니다)

---
## 다중회귀분석 핵심 정리

- **목적** : 여러 독립변수로 연속형 종속변수 예측
- **R²** : 모형이 종속변수 분산을 설명하는 비율 (0~1)
- **Adj. R²** : 변수 개수 페널티 적용
- **`statsmodels.formula.api.ols`** 사용

```python
import numpy as np
import pandas as pd
import statsmodels.formula.api as smf

# 데이터 생성
df = pd.read_csv(DATA_PATH + "sales_5_regression.csv", encoding="utf-8-sig")

# 풀이
formula = "monthly_sales ~ call_count + visit_count + work_years + training_score"
model = smf.ols(formula=formula, data=df).fit()
print(model.summary())

# (1) 회귀계수가 가장 높은 변수
params = model.params.drop("Intercept", errors="ignore")
max_var = params.idxmax()
print("(1) 각 독립변수의 회귀계수:")
for var, coef in params.items():
    print(f"    {var}: {round(float(coef), 4)}")
print(f"\n(1) 회귀계수 최대 변수: {max_var} ({round(float(params[max_var]), 4)})")

# (2) 유의수준 5%에서 유의미한 변수 개수
pvals = model.pvalues.drop("Intercept", errors="ignore")
print("\n(2) 각 독립변수의 p-value:")
for var, pv in pvals.items():
    sig = "유의" if pv < 0.05 else "비유의"
    print(f"    {var}: {round(float(pv), 4)}  [{sig}]")
sig_count = int((pvals < 0.05).sum())
print(f"\n(2) 유의미한 변수 개수: {sig_count}개")

# (3) 결정계수
r2     = round(float(model.rsquared), 3)
adj_r2 = round(float(model.rsquared_adj), 3)
print(f"\n(3) R²       = {r2}")
print(f"(3) Adj. R²  = {adj_r2}")
```

# 문제 6 - 로지스틱 회귀분석

## [문제 6] 대기업 공채 서류 합격 여부 예측 모형 구축 *(HR)*

취업 플랫폼에서 대기업 공채에 지원한 250명의 데이터를 수집하여 **서류 합격(1) 여부**를 예측하는 로지스틱 회귀모형을 구축하고자 한다.

**[데이터]**
- 파일명: `hr_6_logistic.csv`

**[컬럼 설명]**
- `applicant_id` : 지원자 식별 번호
- `univ_tier` : 대학 구분 (0=기타, 1=지방거점국립, 2=수도권상위)
- `toeic_score` : 토익 점수 (점)
- `gpa` : 학점 (4.5 만점)
- `intern_count` : 인턴 경험 횟수 (0~4회)
- `cert_count` : 보유 자격증 수 (0~5개)
- `doc_pass` : 서류 합격 여부 (1=합격, 0=불합격, 종속변수)

**[문항]**
(1) `toeic_score`, `gpa`, `intern_count`, `cert_count`를 독립변수로 하는 로지스틱 회귀모형을 적합하고, **회귀계수가 가장 높은 변수**를 구하시오.
(2) 유의수준 5%에서 **통계적으로 유의미한 변수의 개수**를 구하시오.
(3) 모형의 **정확도(Accuracy)**를 구하시오. (소수점 셋째 자리까지 반올림)

---
## 로지스틱 회귀분석 핵심 정리

- **목적** : 이진 분류(0/1) 종속변수의 발생 확률 예측
- 회귀계수 beta > 0 : 독립변수 증가 시 합격 확률 증가
- 예측 확률 >= 0.5 → 합격(1), 미만 → 불합격(0)
- **`smf.logit(formula, data).fit()`** 사용

```python
import numpy as np
import pandas as pd
import statsmodels.formula.api as smf
from sklearn.metrics import accuracy_score

# 데이터 생성
df = pd.read_csv(DATA_PATH + "hr_6_logistic.csv",encoding="utf-8-sig")

# 풀이
formula = "doc_pass ~ toeic_score + gpa + intern_count + cert_count"
model = smf.logit(formula=formula, data=df).fit(disp=False)

# (1) 회귀계수가 가장 높은 변수
params = model.params.drop("Intercept", errors="ignore")
max_var = params.idxmax()
print("(1) 각 독립변수의 회귀계수 (log-odds):")
for var, coef in params.items():
    print(f"    {var}: {round(float(coef), 4)}")
print(f"\n(1) 회귀계수 최대 변수: {max_var} ({round(float(params[max_var]), 4)})")

# (2) 유의수준 5%에서 유의미한 변수 개수
pvals = model.pvalues.drop("Intercept", errors="ignore")
print("\n(2) 각 독립변수의 p-value:")
for var, pv in pvals.items():
    sig = "유의" if pv < 0.05 else "비유의"
    print(f"    {var}: {round(float(pv), 4)}  [{sig}]")
sig_count = int((pvals < 0.05).sum())
print(f"\n(2) 유의미한 변수 개수: {sig_count}개")

# (3) 모형 정확도
prob_pred = model.predict(df)
y_pred    = (prob_pred >= 0.5).astype(int)
y_true    = df["doc_pass"]
acc       = round(float(accuracy_score(y_true, y_pred)), 3)
print(f"\n(3) 모형 정확도 (Accuracy) = {acc}")
```

# 종합문제 1 - 구직자 역량 데이터 복합 분석 *(HR)*

## [종합문제 1] 취업 준비생 역량 지표 다각도 분석

취업 지원 플랫폼에서 대기업·중견기업·스타트업 최종 합격자 및 불합격자 180명의 데이터를 수집하였다.  
지원 기업 유형별 역량 차이 분석과 최종 합격 예측 모형을 구축하고자 한다.

**[데이터]**
- 파일명: `hr_7_comprehensive1.csv`

**[컬럼 설명]**
- `applicant_id` : 지원자 식별 번호
- `company_type` : 지원 기업 유형 (대기업 / 중견기업 / 스타트업)
- `age` : 나이 (세)
- `gpa` : 학점 (4.5 만점)
- `toeic_score` : 토익 점수 (점)
- `has_intern` : 인턴 경험 여부 (있음 / 없음)
- `final_pass` : 최종 합격 여부 (1=합격, 0=불합격, 종속변수)

**[문항]**
(1) 기업 유형(company_type)별 **토익 점수 평균**을 구하고, One-Way ANOVA로 **F-통계량**과 **p-value**를 구하시오. (소수점 셋째 자리까지 반올림)

(2) 인턴 경험 여부(has_intern)와 최종 합격 여부(final_pass) 간의 연관성을 **카이제곱 검정**으로 분석하고, **chi2-통계량**과 **p-value**를 구하시오. (소수점 셋째 자리까지 반올림)

(3) `age`, `gpa`, `toeic_score`를 독립변수로, `final_pass`를 종속변수로 하는 **로지스틱 회귀모형**을 구축하고, 모형의 **정확도(Accuracy)**와 **AUC**를 구하시오. (소수점 셋째 자리까지 반올림)

```python
import numpy as np
import pandas as pd
from scipy import stats
from scipy.stats import chi2_contingency
import statsmodels.formula.api as smf
from sklearn.metrics import accuracy_score, roc_auc_score

# 데이터 생성
df = pd.read_csv(DATA_PATH + "hr_7_comprehensive1.csv", encoding="utf-8-sig")
print("데이터 shape:", df.shape)
print(df.head(3), "\n")

# (1) 기업 유형별 토익 점수 ANOVA
print("=" * 50)
print("(1) 기업 유형별 토익 점수 분석")
print("=" * 50)
groups = df.groupby("company_type")["toeic_score"]
print("기업 유형별 토익 평균:")
for name, grp in groups:
    print(f"    {name}: {round(float(grp.mean()), 3)} 점")
f_stat, f_p = stats.f_oneway(*[g.values for _, g in groups])
print(f"\nF-통계량: {round(float(f_stat), 3)}")
print(f"p-value : {round(float(f_p), 3)}")
print("결론:", "기각 -> 기업 유형별 토익 점수 차이 있음" if f_p < 0.05 else "채택 -> 차이 없음")

# (2) 인턴 경험 x 최종 합격 카이제곱 검정
print("\n" + "=" * 50)
print("(2) 인턴 경험 여부 x 최종 합격 여부 카이제곱 검정")
print("=" * 50)
ct2 = pd.crosstab(df["has_intern"], df["final_pass"])
print("교차표:")
print(ct2)
chi2_stat, chi_p, dof, _ = chi2_contingency(ct2)
print(f"\nchi2-통계량: {round(float(chi2_stat), 3)}")
print(f"자유도     : {dof}")
print(f"p-value    : {round(float(chi_p), 3)}")
print("결론:", "기각 -> 인턴 경험과 합격 여부에 연관성 있음" if chi_p < 0.05 else "채택 -> 연관성 없음")

# (3) 로지스틱 회귀 + 정확도 + AUC
print("\n" + "=" * 50)
print("(3) 최종 합격 여부 로지스틱 회귀 모형")
print("=" * 50)
model3 = smf.logit("final_pass ~ age + gpa + toeic_score", data=df).fit(disp=False)
print("회귀계수 (log-odds):")
for var, coef in model3.params.drop("Intercept", errors="ignore").items():
    pv  = model3.pvalues[var]
    sig = "유의" if pv < 0.05 else "비유의"
    print(f"    {var}: {round(float(coef), 4)}  (p={round(float(pv), 4)}) [{sig}]")
prob3   = model3.predict(df)
y_pred3 = (prob3 >= 0.5).astype(int)
acc3    = round(float(accuracy_score(df["final_pass"], y_pred3)), 3)
auc3    = round(float(roc_auc_score(df["final_pass"], prob3)), 3)
print(f"\n정확도 (Accuracy) = {acc3}")
print(f"AUC               = {auc3}")
```

# 종합문제 2 - 마케팅 활동이 매출에 미치는 영향 분석 *(세일즈·마케팅)*

## [종합문제 2] 마케팅 투자·고객 만족·SNS 활동이 매출에 미치는 영향

소비재 브랜드 마케팅팀에서 영업 담당자 220명을 대상으로 **마케팅 예산 집행액**, **고객 만족도**, **SNS 팔로워 증가율**, **리텐션율**을 조사하고, **월매출(만 원)**과의 관계를 분석하고자 한다.

**[데이터]**
- 파일명: `sales_8_comprehensive2.csv`

**[컬럼 설명]**
- `emp_id` : 직원 식별 번호
- `tier` : 영업 등급 (Bronze / Silver / Gold / Platinum)
- `region` : 영업 지역 (수도권 / 강원 / 충청 / 경상 / 전라)
- `mkt_budget` : 월간 마케팅 예산 집행액 (만 원)
- `csat_score` : 고객 만족도 점수 (1~5, 리커트 척도)
- `sns_growth` : SNS 팔로워 증가율 (%)
- `retention_rate` : 고객 리텐션율 (%)
- `monthly_sales` : 월매출 (만 원, 종속변수)

**[문항]**
(1) `mkt_budget`, `csat_score`, `sns_growth`, `retention_rate`를 독립변수로 하는 **다중회귀모형**을 구축하고,  
유의수준 5%에서 **유의미한 변수의 이름과 개수**를 구하고, 회귀모형의 **R²**를 구하시오. (소수점 셋째 자리까지 반올림)

(2) 영업 등급(tier)에 따른 **월매출 평균**이 동일한지 **One-Way ANOVA**로 검정하고,  
**F-통계량**, **p-value**, 귀무가설 **기각/채택** 여부를 구하시오. (소수점 셋째 자리까지 반올림)

(3) SNS 팔로워 증가율(sns_growth)과 고객 만족도(csat_score)의 **피어슨 상관계수**를 구하고,  
유의수준 5%에서 통계적으로 유의미한 상관관계인지 판단하시오. (소수점 셋째 자리까지 반올림)

```python
import numpy as np
import pandas as pd
from scipy import stats
import statsmodels.formula.api as smf

# 데이터 생성
df = pd.read_csv(DATA_PATH + "sales_8_comprehensive2.csv", encoding="utf-8-sig")
print("데이터 shape:", df.shape)
print(df.head(3), "\n")

# (1) 다중회귀분석
print("=" * 50)
print("(1) 월매출 예측 다중회귀분석")
print("=" * 50)
formula = "monthly_sales ~ mkt_budget + csat_score + sns_growth + retention_rate"
model1  = smf.ols(formula, data=df).fit()
params  = model1.params.drop("Intercept", errors="ignore")
pvals   = model1.pvalues.drop("Intercept", errors="ignore")
print("변수별 회귀계수 및 유의성:")
sig_vars = []
for var in params.index:
    coef = params[var]
    pv   = pvals[var]
    sig  = "유의" if pv < 0.05 else "비유의"
    if pv < 0.05:
        sig_vars.append(var)
    print(f"    {var}: beta={round(float(coef), 4)}, p={round(float(pv), 4)}  [{sig}]")
r2 = round(float(model1.rsquared), 3)
print(f"\n유의미한 변수: {sig_vars} ({len(sig_vars)}개)")
print(f"R² = {r2}")

# (2) 영업 등급별 월매출 ANOVA
print("\n" + "=" * 50)
print("(2) 영업 등급별 월매출 ANOVA")
print("=" * 50)
tier_order = ["Bronze", "Silver", "Gold", "Platinum"]
groups2    = [df[df["tier"] == t]["monthly_sales"].values for t in tier_order]
print("등급별 월매출 평균:")
for t, g in zip(tier_order, groups2):
    print(f"    {t}: {round(float(g.mean()), 3)} 만 원")
f2, p2 = stats.f_oneway(*groups2)
print(f"\nF-통계량: {round(float(f2), 3)}")
print(f"p-value : {round(float(p2), 3)}")
decision2 = "기각" if p2 < 0.05 else "채택"
print(f"귀무가설 (H0: 등급별 월매출 평균 동일) -> {decision2}")

# (3) 피어슨 상관계수
print("\n" + "=" * 50)
print("(3) sns_growth x csat_score 피어슨 상관계수")
print("=" * 50)
r_corr, r_p = stats.pearsonr(df["sns_growth"], df["csat_score"])
r_corr = round(float(r_corr), 3)
r_p    = round(float(r_p), 3)
print(f"피어슨 r = {r_corr}")
print(f"p-value  = {r_p}")
corr_decision = "유의미한 상관관계 있음" if r_p < 0.05 else "유의미한 상관관계 없음"
direction     = "부적 상관" if r_corr < 0 else "정적 상관"
print(f"결론: {corr_decision} ({direction})")
```
