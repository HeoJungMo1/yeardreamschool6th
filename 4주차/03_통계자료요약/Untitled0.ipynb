```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

# 2행 3열의 서브플롯 생성
fig, ax = plt.subplots(2, 3, figsize=(12, 6))

# 데이터 생성
x = np.linspace(0, 10, 100)

# 각 서브플롯에 접근하여 그래프 그리기
for i in range(2):
    for j in range(3):
        ax[i, j].plot(x, np.sin(x + (i + j)))
        ax[i, j].set_title(f'Subplot {i+1},{j+1}')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 데이터 로드 (예시)
iris = sns.load_dataset("iris")

# 2행 3열 그리드 생성
fig, axes = plt.subplots(2, 3, figsize=(14, 8))
fig.suptitle("iris dataset — fig, ax pattern demo", fontsize=16)

# 1행 (0번 인덱스)
sns.scatterplot(data=iris, x="sepal_length", y="petal_length", hue="species", ax=axes[0, 0])
sns.violinplot(data=iris, x="species", y="sepal_width", ax=axes[0, 1])
sns.boxplot(data=iris, x="species", y="petal_width", ax=axes[0, 2])

# 2행 (1번 인덱스)
sns.histplot(data=iris, x="sepal_length", hue="species", multiple="stack", ax=axes[1, 0])
sns.kdeplot(data=iris, x="sepal_width", hue="species", fill=True, ax=axes[1, 1])
sns.barplot(data=iris, x="species", y="sepal_length", ax=axes[1, 2])

# 레이아웃 최적화 및 제목 공간 확보
fig.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```
