## Part 1. Timestamp & 날짜 생성

> `pd.Timestamp`, `pd.to_datetime`, `pd.date_range`, `pd.period_range`


### Q01. 문자열 -> Timestamp 변환 (3가지 방법)

```python
# 방법 1 : pd.Timestamp
ts1 = pd.Timestamp("2026-07-01")
ts1

# 방법 2 : pd.to_datetime (단일 문자열)
ts2 = pd.to_datetime('2026-01-06')
ts2

# 방법 3 : datetime 객체
ts3 = pd.Timestamp(datetime.datetime(2026, 7, 1))

print(type(ts1))
print(ts1)
print(type(ts2))
print(ts2)
print(type(ts3))
print(ts3)
```

### Q02. 다양한 형식의 날짜 문자열을 한번에 변환

```python
raw_dates = ['2003/01/06', 'Jan 9, 2003', '2003-01-10', '10-01-2003']

parsed = pd.to_datetime(raw_dates, dayfirst = False, format = 'mixed')
print(parsed)
```

### Q03. 잘못된 날짜 처리 — errors='coerce'

```python
messy = orders['orderDate'].astype(str).tolist() + ['NOT_A_DATE', '9999-99-99']
parsed = pd.to_datetime(messy, errors = 'coerce')
print(parsed)

# 날짜 하나씩 처리 할 때는 TimeStamp 객체로 반환
# 날짜를 묶음으로 관리할 때는 DatetimeIndex 객체로 반환 (datetime64[ns])
```

### Q04. Unix epoch(초) -> Timestamp 변환

```python
# 주문일을 epoch초로 변환을 강제로 시킴
# 다시 복원
# 첫 번째, 날짜기리 연산이 됨
# 두 번째 날짜 데이터끼리 연산이 되면, days, 초 형태로 반환이 됨

epoch_seconds = (orders['orderDate'] - pd.Timestamp('1970-01-01')) // pd.Timedelta('1s')
# epoch_seconds

restored = pd.to_datetime(epoch_seconds, unit="s")
restored
```

### Q05. 여러 컬럼(year, month, day)으로 날짜 조립

## Part 2. .dt 접근자 — 시분초 분해

> Series 에 datetime dtype 이 있으면 `.dt` 로 날짜 구성 요소를 추출할 수 있다.


### Q11. .dt.year / .dt.month / .dt.day 추출

```python
orders['year'] = orders['orderDate'].dt.year
orders['month'] = orders['orderDate'].dt.month
orders['day'] = orders['orderDate'].dt.day
orders.head()
```

### Q12. .dt.hour / .dt.minute / .dt.second 추출 (시분초)

```python
# 시분초 테스트용 데이터 생성
ts_series = pd.Series(pd.date_range("2003-01-06 09:05:30", periods=5, freq="90min"))
ts_series.dt.second.tolist()
ts_series.dt.hour.tolist()
ts_series.dt.minute.tolist()
```

### Q13. .dt.day_name() — 요일 이름 추출

```python
a = orders['weekday'] = orders['orderDate'].dt.day_name()
print(a)
orders['weekday'].value_counts()
```

### Q14. .dt.dayofweek — 요일 번호 (월=0 ~ 일=6)

```python
orders['dow'] = orders['orderDate'].dt.dayofweek
# 월 0, 토 5, 일 6
orders['is_weekend'] = orders['dow'] >= 5
orders.head()
```

```python
orders['dow'] >= 5
```

### Q15. .dt.quarter — 분기 추출

```python
orders['quarter'] = orders['orderDate'].dt.quarter
orders.groupby('quarter')['orderNumber'].count()
```

### Q16. .dt.week / .dt.isocalendar() — ISO 주 번호

```python
iso = orders['orderDate'].dt.isocalendar()
orders['iso_week'] = iso['week']
orders.groupby('iso_week')['orderNumber'].count()
```

### Q17. .dt.days_in_month — 해당 월의 총 일수

```python
payments['paymentDate'].dt.day
# 2026-06-30 => 30
# dt.days_in_month => 그 달의 총 일수
```

```python
payments['days_in_month'] = payments['paymentDate'].dt.days_in_month
payments['is_last_day'] = (payments['paymentDate'].dt.day == payments['paymentDate'].dt.days_in_month)
payments
```

### Q18. .dt.normalize() — 시간 부분을 00:00:00 으로 초기화

```python
ts_series = pd.Series(pd.date_range("2003-01-06 09:05:30", periods=4, freq="90min"))
print("원본:")
print(ts_series)
print("\nnormalize (시간 제거):")
print(ts_series.dt.normalize())
```

### Q19. .dt.strftime() — 날짜 -> 원하는 문자열 포맷

```python
orders['date_str_kr'] = orders['orderDate'].dt.strftime('%m월 %Y년 %d일')
orders['date_str_kr'] = orders['orderDate'].dt.strftime('%B %d %Y')
orders.head()
```

### Q32. 커스텀 영업일 — 공휴일 제외

```python
# 한국 공휴일 일부 설정
hr_holidays = [
    pd.Timestamp('2003-01-01'), # 신정
    pd.Timestamp('2003-03-01'), # 삼일절
    pd.Timestamp('2003-05-05'), # 어린이날
    pd.Timestamp('2003-07-17') # 제헌절
]

# 아래 문법 기억하자
KR_BDay = pd.offsets.CustomBusinessDay(holidays = hr_holidays)
KR_BDay

ts = pd.Timestamp('2002-12-28') # 토요일
print(ts, ts.day_name())

new_year = pd.Timestamp('2003-01-01')
print('신정 다음 한국 영업일 : ', new_year + KR_BDay)
```
