```python
class TeamMember:
    # ── 클래스 변수: 모든 팀원이 같은 값을 공유 ──
    company_name = "TechCorp"
    total_members = 0

    def __init__(self, name, role, salary):
        self.name = name
        self.role = role
        self.salary = salary

        TeamMember.total_members += 1

    def introduce(self):
        print(f"[{TeamMember.company_name}] 이름 : {self.name} 직책 : {self.role}, 연봉 : {self.salary}만원")
        

carol = TeamMember(name = "carol", role = "주니어 개발자", salary = 5000)
alice = TeamMember(name = "alice", role = "시니어 개발자", salary = 7000)
bob = TeamMember(name = "bob", role = "주니어 개발자", salary = 4500)

alice.introduce()
bob.introduce()
carol.introduce()
```
